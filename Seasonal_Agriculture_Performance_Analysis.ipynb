Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load the Dataset

```python
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

Check Missing Values

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values[missing_values > 0])
```

Handle Missing Values Safely

```python
# Numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum().sum())
```

Check and Remove Duplicate Rows

```python
print("Duplicate rows before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("New dataset shape:", df.shape)
```

Season-wise Number of Records

```python
season_counts = df["Season"].value_counts()

print(season_counts)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Season")

plt.title("Number of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")

plt.show()
```

```python
avg_yield = (
    df.groupby("Season")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
)

print(avg_yield)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=avg_yield.index,
    y=avg_yield.values
)

plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Hectare)")

plt.show()
```

Find Top Performing Season

```python
season_summary = df.groupby("Season").agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Rainfall=("Rainfall_mm", "mean"),
    Average_Temperature=("Avg_Temperature_C", "mean"),
    Average_Water_Usage=("Water_Used_m3", "mean")
)

season_summary = season_summary.round(2)

print(season_summary)
```

Best Season for Each Metric

```python
print(
    "Best season by average yield:",
    season_summary["Average_Yield"].idxmax()
)

print(
    "Best season by total production:",
    season_summary["Total_Production"].idxmax()
)

print(
    "Best season by average profit:",
    season_summary["Average_Profit"].idxmax()
)
```

Top 10 Most Profitable Crops

```python
top_profitable_crops = (
    df.groupby("Crop")["Profit_INR"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(top_profitable_crops)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_profitable_crops.values,
    y=top_profitable_crops.index
)

plt.title("Top Crops by Average Profit")
plt.xlabel("Average Profit (INR)")
plt.ylabel("Crop")

plt.show()
```
